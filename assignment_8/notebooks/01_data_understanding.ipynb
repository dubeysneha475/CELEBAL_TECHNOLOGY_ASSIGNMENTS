## 2.11 Validate Item Total

The `item_total` is the total price for each individual product.

I will validate whether:

item_total = quantity × item_price

This test will allow us to detect mistakes in our raw data.

I will not make any changes to the initial raw data at this stage.

```python
# ============================================================
# STEP 2.11 - VALIDATE ITEM TOTAL
# ============================================================

# Calculate the expected item total
# without modifying the original order_items DataFrame

calculated_item_total = (
    order_items["quantity"] *
    order_items["item_price"]
)


# Calculate the difference between:
# Existing item_total
# and
# Calculated item total

total_difference = (
    order_items["item_total"] -
    calculated_item_total
)


# Find rows where the difference is greater than 0.01
# We use 0.01 because prices are generally measured in decimals.

incorrect_item_totals = order_items[
    ~np.isclose(
        order_items["item_total"],
        calculated_item_total,
        atol=0.01
    )
]


# Display the number of inconsistent records

print(
    "Rows with inconsistent item_total:",
    len(incorrect_item_totals)
)
```

- Product price cannot be negative.

```python
# ============================================================
# STEP 4.7 - FIX INVALID NUMERIC VALUES
# ============================================================

# ------------------------------------------------------------
# 1. Fix invalid quantity
# ------------------------------------------------------------

# Identify valid quantities
valid_quantities = order_items_clean[
    order_items_clean["quantity"] > 0
]["quantity"]

# Calculate median of valid quantities
quantity_median = valid_quantities.median()

# Replace quantity <= 0 with median
order_items_clean.loc[
    order_items_clean["quantity"] <= 0,
    "quantity"
] = quantity_median


# ------------------------------------------------------------
# 2. Fix negative item prices
# ------------------------------------------------------------

# Identify valid item prices
valid_item_prices = order_items_clean[
    order_items_clean["item_price"] >= 0
]["item_price"]

# Calculate median valid price
item_price_median = valid_item_prices.median()

# Replace negative prices
order_items_clean.loc[
    order_items_clean["item_price"] < 0,
    "item_price"
] = item_price_median


# ------------------------------------------------------------
# 3. Fix negative product prices
# ------------------------------------------------------------

# Identify valid product prices
valid_product_prices = products_clean[
    products_clean["price"] >= 0
]["price"]

# Calculate median valid product price
product_price_median = valid_product_prices.median()

# Replace negative prices
products_clean.loc[
    products_clean["price"] < 0,
    "price"
] = product_price_median


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

print(
    "Invalid quantities remaining:",
    (order_items_clean["quantity"] <= 0).sum()
)

print(
    "Negative item prices remaining:",
    (order_items_clean["item_price"] < 0).sum()
)

print(
    "Negative product prices remaining:",
    (products_clean["price"] < 0).sum()
)
```

## 4.8 Recalculate Item Total

The `item_total` column may contain incorrect values due to data-entry
or processing errors.

The correct item total is calculated using:

item_total = quantity × item_price

Recalculating this field ensures consistency between quantity, price,
and total amount.

```python
# ============================================================
# STEP 4.8 - RECALCULATE ITEM TOTAL
# ============================================================

# Calculate the correct item total
order_items_clean["calculated_item_total"] = (
    order_items_clean["quantity"]
    * order_items_clean["item_price"]
)


# ------------------------------------------------------------
# Find records where the existing total is incorrect
# ------------------------------------------------------------

incorrect_item_totals = order_items_clean[
    ~np.isclose(
        order_items_clean["item_total"],
        order_items_clean["calculated_item_total"],
        rtol=1e-05,
        atol=0.01
    )
]


print(
    "Incorrect item_total records found:",
    len(incorrect_item_totals)
)


# ------------------------------------------------------------
# Replace incorrect totals with calculated totals
# ------------------------------------------------------------

order_items_clean["item_total"] = (
    order_items_clean["calculated_item_total"]
)


# ------------------------------------------------------------
# Remove temporary calculation column
# ------------------------------------------------------------

order_items_clean = order_items_clean.drop(
    columns=["calculated_item_total"]
)


# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

validation_check = np.isclose(
    order_items_clean["item_total"],
    order_items_clean["quantity"]
    * order_items_clean["item_price"],
    rtol=1e-05,
    atol=0.01
)


print(
    "Incorrect item_total records remaining:",
    (~validation_check).sum()
)
```

- Average order value

# 8. SQL Window Functions

Window functions allow us to perform calculations across related rows
without grouping the rows into a single result.

In this section, we rank customers within each city based on their
total spending.

This helps identify the highest-value customers in different cities.

```python
# ============================================================
# STEP 8.1 - RANK CUSTOMERS WITH WINDOW FUNCTION
# ============================================================

top_customers_by_city = pd.read_sql_query(
    """

    -- First calculate total spending for every customer
    WITH customer_spending AS (

        SELECT
            u.user_id,
            u.name,
            u.city,

            -- Calculate total money spent
            ROUND(SUM(oi.item_total), 2) AS total_spent

        FROM users AS u

        -- Connect users to orders
        INNER JOIN orders AS o
            ON u.user_id = o.user_id

        -- Connect orders to order items
        INNER JOIN order_items AS oi
            ON o.order_id = oi.order_id

        GROUP BY
            u.user_id,
            u.name,
            u.city
    )


    -- Rank customers inside each city
    SELECT
        user_id,
        name,
        city,
        total_spent,

        RANK() OVER (
            PARTITION BY city
            ORDER BY total_spent DESC
        ) AS city_rank

    FROM customer_spending

    ORDER BY
        city,
        city_rank;

    """,
    connection
)

display(top_customers_by_city.head(20))
```

# 9. Customer Segmentation Using CTE

Customers are segmented according to their total spending.

Segments:

- High Value: Total spending >= 5000
- Medium Value: Total spending between 2000 and 4999.99
- Low Value: Total spending < 2000

A Common Table Expression (CTE) is used to first calculate customer
spending and then classify customers into segments.

```python
# ============================================================
# STEP 9.1 - CUSTOMER SEGMENTATION USING CTE
# ============================================================

customer_segments = pd.read_sql_query(
    """

    -- ========================================================
    -- CTE: Calculate total spending for each customer
    -- ========================================================

    WITH customer_spending AS (

        SELECT
            u.user_id,
            u.name,
            u.city,

            -- Total amount spent by the customer
            ROUND(SUM(oi.item_total), 2) AS total_spent,

            -- Number of orders made by the customer
            COUNT(DISTINCT o.order_id) AS total_orders

        FROM users AS u

        INNER JOIN orders AS o
            ON u.user_id = o.user_id

        INNER JOIN order_items AS oi
            ON o.order_id = oi.order_id

        GROUP BY
            u.user_id,
            u.name,
            u.city
    )


    -- ========================================================
    -- Segment customers based on total spending
    -- ========================================================

    SELECT
        user_id,
        name,
        city,
        total_orders,
        total_spent,

        CASE

            -- Customers spending 5000 or more
            WHEN total_spent >= 5000
                THEN 'High Value'

            -- Customers spending between 2000 and 4999.99
            WHEN total_spent >= 2000
                THEN 'Medium Value'

            -- Customers spending less than 2000
            ELSE 'Low Value'

        END AS customer_segment

    FROM customer_spending

    ORDER BY
        total_spent DESC;

    """,
    connection
)

display(customer_segments.head(20))
```

## 9.2 Customer Segment Distribution

The number of customers in each segment is calculated to understand
the distribution of customer value.

```python
# ============================================================
# STEP 9.2 - SEGMENT DISTRIBUTION
# ============================================================

segment_distribution = pd.read_sql_query(
    """

    WITH customer_spending AS (

        SELECT
            u.user_id,

            ROUND(SUM(oi.item_total), 2) AS total_spent

        FROM users AS u

        INNER JOIN orders AS o
            ON u.user_id = o.user_id

        INNER JOIN order_items AS oi
            ON o.order_id = oi.order_id

        GROUP BY
            u.user_id
    )


    SELECT

        CASE

            WHEN total_spent >= 5000
                THEN 'High Value'

            WHEN total_spent >= 2000
                THEN 'Medium Value'

            ELSE 'Low Value'

        END AS customer_segment,

        COUNT(*) AS customer_count,

        ROUND(
            SUM(total_spent),
            2
        ) AS segment_revenue

    FROM customer_spending

    GROUP BY
        customer_segment

    ORDER BY
        segment_revenue DESC;

    """,
    connection
)

display(segment_distribution)
```

# 10. Cohort Analysis and Customer Retention

Cohort analysis groups customers based on the month of their first order.

We then track whether these customers placed additional orders in
subsequent months.

This helps measure customer retention over time.

```python
# ============================================================
# STEP 10.1 - CUSTOMER COHORT ANALYSIS
# ============================================================

cohort_analysis = pd.read_sql_query(
    """

    -- ========================================================
    -- CTE 1: Find the first order month for every customer
    -- ========================================================

    WITH customer_first_order AS (

        SELECT
            user_id,

            -- Find the month of customer's first order
            MIN(
                date(order_date)
            ) AS first_order_date

        FROM orders

        GROUP BY
            user_id
    ),


    -- ========================================================
    -- CTE 2: Get every customer's order month
    -- ========================================================

    customer_orders AS (

        SELECT
            o.user_id,

            date(o.order_date) AS order_date,

            c.first_order_date

        FROM orders AS o

        INNER JOIN customer_first_order AS c

            ON o.user_id = c.user_id
    )


    -- ========================================================
    -- Final cohort result
    -- ========================================================

    SELECT

        -- Month when the customer first purchased
        strftime(
            '%Y-%m',
            first_order_date
        ) AS cohort_month,

        -- Month when the customer made a purchase
        strftime(
            '%Y-%m',
            order_date
        ) AS order_month,

        -- Number of unique customers
        COUNT(
            DISTINCT user_id
        ) AS active_customers

    FROM customer_orders

    GROUP BY

        cohort_month,
        order_month

    ORDER BY

        cohort_month,
        order_month;

    """,
    connection
)

display(cohort_analysis.head(30))
```

## 10.2 Calculate Months Since First Purchase

The retention month represents the number of months between the customer's
first purchase and a later purchase.

For example:

- Month 0 = First purchase
- Month 1 = Returned one month later
- Month 2 = Returned two months later
- Month 3 = Returned three months later

```python
# ============================================================
# STEP 10.2 - RETENTION MONTH CALCULATION
# ============================================================

retention_analysis = pd.read_sql_query(
    """

    WITH customer_first_order AS (

        SELECT
            user_id,

            MIN(
                date(order_date)
            ) AS first_order_date

        FROM orders

        GROUP BY
            user_id
    ),


    customer_orders AS (

        SELECT
            o.user_id,

            date(o.order_date) AS order_date,

            c.first_order_date

        FROM orders AS o

        INNER JOIN customer_first_order AS c

            ON o.user_id = c.user_id
    )


    SELECT

        -- Customer's first order month
        strftime(
            '%Y-%m',
            first_order_date
        ) AS cohort_month,


        -- Month of the customer's order
        strftime(
            '%Y-%m',
            order_date
        ) AS order_month,


        -- Calculate months between first and current order
        (
            (
                CAST(
                    strftime('%Y', order_date)
                    AS INTEGER
                )
                -
                CAST(
                    strftime('%Y', first_order_date)
                    AS INTEGER
                )
            ) * 12
            +
            (
                CAST(
                    strftime('%m', order_date)
                    AS INTEGER
                )
                -
                CAST(
                    strftime('%m', first_order_date)
                    AS INTEGER
                )
            )
        ) AS retention_month,


        -- Number of active customers
        COUNT(
            DISTINCT user_id
        ) AS active_customers


    FROM customer_orders


    GROUP BY

        cohort_month,
        order_month,
        retention_month


    ORDER BY

        cohort_month,
        retention_month;

    """,
    connection
)

display(retention_analysis.head(30))
```

```python
# ============================================================
# COHORT RETENTION RATE
# ============================================================

# Find the number of customers in each cohort's first month
# This will be our base customer count.
cohort_size = (
    retention_analysis[
        retention_analysis["retention_month"] == 0
    ][["cohort_month", "active_customers"]]
    .rename(columns={"active_customers": "cohort_size"})
)

# Merge cohort size with the retention analysis
retention_rate = retention_analysis.merge(
    cohort_size,
    on="cohort_month",
    how="left"
)

# Calculate retention percentage
retention_rate["retention_rate"] = (
    retention_rate["active_customers"]
    / retention_rate["cohort_size"]
    * 100
)

# Create a cohort retention matrix
cohort_matrix = retention_rate.pivot(
    index="cohort_month",
    columns="retention_month",
    values="retention_rate"
)

# Round percentages to 2 decimal places
cohort_matrix = cohort_matrix.round(2)

print("COHORT RETENTION RATE (%)")
display(cohort_matrix)
```

```python
# ============================================================
# CUSTOMER SEGMENTATION ANALYSIS
# ============================================================

# Create customer segments based on total spending
# Low Value    : less than 500
# Medium Value : 500 to 2000
# High Value   : more than 2000

customer_summary["customer_segment"] = pd.cut(
    customer_summary["total_spent"],
    bins=[-np.inf, 500, 2000, np.inf],
    labels=["Low Value", "Medium Value", "High Value"]
)

# Count customers and calculate revenue for each segment
segment_distribution = (
    customer_summary
    .groupby("customer_segment", observed=False)
    .agg(
        customer_count=("user_id", "count"),
        segment_revenue=("total_spent", "sum")
    )
    .reset_index()
)

# Round revenue
segment_distribution["segment_revenue"] = (
    segment_distribution["segment_revenue"].round(2)
)

print("CUSTOMER SEGMENT DISTRIBUTION")
display(segment_distribution)
```

```python
# ============================================================
# RFM ANALYSIS
# ============================================================
# RFM stands for:
# R = Recency    -> How recently the customer purchased
# F = Frequency  -> How many orders the customer made
# M = Monetary   -> How much the customer spent
# ============================================================

# Make sure order_date is in datetime format
orders_clean["order_date"] = pd.to_datetime(orders_clean["order_date"])

# Set the reference date as one day after the latest order
# This makes the recency calculation consistent with our dataset.
reference_date = orders_clean["order_date"].max() + pd.Timedelta(days=1)

# Calculate RFM values for every customer
rfm_analysis = (
    orders_clean
    .groupby("user_id")
    .agg(
        # Recency: days since the customer's most recent order
        recency=("order_date", lambda x:
                 (reference_date - x.max()).days),

        # Frequency: total number of orders
        frequency=("order_id", "nunique"),

        # Monetary: total amount spent
        monetary=("total_amount", "sum")
    )
    .reset_index()
)

# Round monetary value to 2 decimal places
rfm_analysis["monetary"] = rfm_analysis["monetary"].round(2)

print("RFM ANALYSIS")
display(rfm_analysis.head(10))
```

```python
# ============================================================
# RFM SCORING
# ============================================================
# We assign each customer a score from 1 to 5.
#
# Recency:
#   Lower recency = better customer
#
# Frequency:
#   Higher frequency = better customer
#
# Monetary:
#   Higher spending = better customer
# ============================================================

# Recency score
# Recent customers get a higher score.
rfm_analysis["R_score"] = pd.qcut(
    rfm_analysis["recency"],
    q=5,
    labels=[5, 4, 3, 2, 1],
    duplicates="drop"
).astype(int)

# Frequency score
# Customers with more orders get a higher score.
rfm_analysis["F_score"] = pd.qcut(
    rfm_analysis["frequency"].rank(method="first"),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Monetary score
# Customers who spend more get a higher score.
rfm_analysis["M_score"] = pd.qcut(
    rfm_analysis["monetary"],
    q=5,
    labels=[1, 2, 3, 4, 5],
    duplicates="drop"
).astype(int)

# Create one combined RFM score
rfm_analysis["RFM_score"] = (
    rfm_analysis["R_score"].astype(str)
    + rfm_analysis["F_score"].astype(str)
    + rfm_analysis["M_score"].astype(str)
)

print("RFM SCORES")
display(rfm_analysis.head(10))
```

```python
# ============================================================
# FINAL RFM CUSTOMER SEGMENTS
# ============================================================
# We classify customers based on their combined RFM score.
#
# High Value Customer  -> Strong RFM score
# Medium Value Customer -> Average RFM score
# Low Value Customer   -> Lower RFM score
# ============================================================

# Convert individual scores into a numeric total
rfm_analysis["RFM_total"] = (
    rfm_analysis["R_score"]
    + rfm_analysis["F_score"]
    + rfm_analysis["M_score"]
)

# Create customer segments based on total RFM score
rfm_analysis["customer_segment"] = pd.cut(
    rfm_analysis["RFM_total"],
    bins=[0, 7, 11, 15],
    labels=["Low Value", "Medium Value", "High Value"],
    include_lowest=True
)

# Count customers in each segment
rfm_segments = (
    rfm_analysis
    .groupby("customer_segment", observed=False)
    .agg(
        customer_count=("user_id", "count"),
        total_revenue=("monetary", "sum")
    )
    .reset_index()
)

# Round revenue
rfm_segments["total_revenue"] = rfm_segments["total_revenue"].round(2)

print("FINAL RFM CUSTOMER SEGMENTS")
display(rfm_segments)
```

```python
# ============================================================
# PRODUCT PERFORMANCE ANALYSIS
# ============================================================
# We will calculate:
# 1. Total quantity sold for each product
# 2. Total revenue generated by each product
# 3. Number of orders containing each product
# ============================================================

# Aggregate order item data by product
product_performance = (
    order_items_clean
    .groupby("product_id")
    .agg(
        total_quantity_sold=("quantity", "sum"),
        total_revenue=("item_total", "sum"),
        total_orders=("order_id", "nunique")
    )
    .reset_index()
)

# Add product names and categories
product_performance = product_performance.merge(
    products_clean[["product_id", "product_name", "category", "brand"]],
    on="product_id",
    how="left"
)

# Sort products by total revenue
product_performance = product_performance.sort_values(
    "total_revenue",
    ascending=False
)

# Round revenue
product_performance["total_revenue"] = (
    product_performance["total_revenue"].round(2)
)

print("TOP 10 PRODUCTS BY REVENUE")

display(
    product_performance[
        [
            "product_id",
            "product_name",
            "category",
            "brand",
            "total_quantity_sold",
            "total_orders",
            "total_revenue"
        ]
    ].head(10)
)
```

```python
# ============================================================
# CATEGORY PERFORMANCE ANALYSIS
# ============================================================
# We will calculate:
# 1. Number of products in each category
# 2. Total quantity sold
# 3. Total orders
# 4. Total revenue
# ============================================================

category_performance = (
    product_performance
    .groupby("category")
    .agg(
        product_count=("product_id", "nunique"),
        total_quantity_sold=("total_quantity_sold", "sum"),
        total_orders=("total_orders", "sum"),
        total_revenue=("total_revenue", "sum")
    )
    .reset_index()
)

# Sort categories by revenue
category_performance = category_performance.sort_values(
    "total_revenue",
    ascending=False
)

# Round revenue
category_performance["total_revenue"] = (
    category_performance["total_revenue"].round(2)
)

print("CATEGORY PERFORMANCE")

display(category_performance)
```

```python
# ============================================================
# BRAND PERFORMANCE ANALYSIS
# ============================================================
# We will calculate:
# 1. Number of products for each brand
# 2. Total quantity sold
# 3. Total orders
# 4. Total revenue
# ============================================================

brand_performance = (
    product_performance
    .groupby("brand")
    .agg(
        product_count=("product_id", "nunique"),
        total_quantity_sold=("total_quantity_sold", "sum"),
        total_orders=("total_orders", "sum"),
        total_revenue=("total_revenue", "sum")
    )
    .reset_index()
)

# Sort brands by total revenue
brand_performance = brand_performance.sort_values(
    "total_revenue",
    ascending=False
)

# Round revenue
brand_performance["total_revenue"] = (
    brand_performance["total_revenue"].round(2)
)

print("TOP 10 BRANDS BY REVENUE")

display(brand_performance.head(10))
```

```python
# ============================================
# 16. PRODUCTS FREQUENTLY BOUGHT TOGETHER
# ============================================

# Keep only required columns
pair_data = order_items_clean[['order_id', 'product_id']].copy()

# Remove duplicate products within the same order
pair_data = pair_data.drop_duplicates()

# Self-join on order_id to find products bought in the same order
product_pairs = pair_data.merge(
    pair_data,
    on='order_id'
)

# Remove same-product pairs
product_pairs = product_pairs[
    product_pairs['product_id_x'] != product_pairs['product_id_y']
]

# Keep each pair only once:
# A-B should be included, but B-A should not
product_pairs['product_a'] = product_pairs[['product_id_x', 'product_id_y']].min(axis=1)
product_pairs['product_b'] = product_pairs[['product_id_x', 'product_id_y']].max(axis=1)

# Count how many times each pair was bought together
frequent_pairs = (
    product_pairs
    .groupby(['product_a', 'product_b'])
    .size()
    .reset_index(name='times_bought_together')
    .sort_values('times_bought_together', ascending=False)
)

print("TOP PRODUCTS FREQUENTLY BOUGHT TOGETHER")

display(frequent_pairs.head(20))
```

```python
# ============================================
# STEP 4: SALES SUMMARY REPORT
# ============================================

from datetime import datetime, timedelta

# Convert input dates to date objects
start = datetime.strptime(start_date, "%Y-%m-%d").date()
end = datetime.strptime(end_date, "%Y-%m-%d").date()

# Calculate previous period
period_days = (end - start).days + 1
previous_end = start - timedelta(days=1)
previous_start = previous_end - timedelta(days=period_days - 1)

# ============================================
# CURRENT PERIOD SUMMARY
# ============================================

summary_query = """
SELECT
    COUNT(DISTINCT order_id) AS total_orders,
    COALESCE(SUM(total_amount), 0) AS total_revenue,
    COUNT(DISTINCT user_id) AS unique_customers
FROM orders
WHERE DATE(order_date) BETWEEN ? AND ?
"""

summary = connection.execute(
    summary_query,
    (start_date, end_date)
).fetchone()

total_orders = summary[0]
total_revenue = summary[1]
unique_customers = summary[2]


# ============================================
# TOP 3 PRODUCTS
# ============================================

top_products_query = """
SELECT
    p.product_name,
    SUM(oi.quantity) AS quantity_sold,
    SUM(oi.item_total) AS product_revenue
FROM order_items oi
JOIN orders o
    ON oi.order_id = o.order_id
JOIN products p
    ON oi.product_id = p.product_id
WHERE DATE(o.order_date) BETWEEN ? AND ?
GROUP BY p.product_id, p.product_name
ORDER BY product_revenue DESC
LIMIT 3
"""

top_products = connection.execute(
    top_products_query,
    (start_date, end_date)
).fetchall()


# ============================================
# PREVIOUS PERIOD SUMMARY
# ============================================

previous_query = """
SELECT
    COUNT(DISTINCT order_id) AS total_orders,
    COALESCE(SUM(total_amount), 0) AS total_revenue
FROM orders
WHERE DATE(order_date) BETWEEN ? AND ?
"""

previous = connection.execute(
    previous_query,
    (previous_start.isoformat(), previous_end.isoformat())
).fetchone()

previous_orders = previous[0]
previous_revenue = previous[1]


# ============================================
# CALCULATE % CHANGE
# ============================================

if previous_revenue != 0:
    revenue_change = (
        (total_revenue - previous_revenue)
        / previous_revenue
    ) * 100
else:
    revenue_change = 0

if previous_orders != 0:
    orders_change = (
        (total_orders - previous_orders)
        / previous_orders
    ) * 100
else:
    orders_change = 0


# ============================================
# PRINT FINAL REPORT
# ============================================

print("=" * 50)
print("           SALES SUMMARY REPORT")
print("=" * 50)

print(f"Report Type       : {report_type}")
print(f"Start Date        : {start_date}")
print(f"End Date          : {end_date}")

print("\nSUMMARY")
print(f"Total Orders      : {total_orders}")
print(f"Total Revenue     : {total_revenue:.2f}")
print(f"Unique Customers  : {unique_customers}")

print("\nTOP 3 PRODUCTS")

for i, product in enumerate(top_products, start=1):
    print(
        f"{i}. {product[0]} "
        f"| Quantity Sold: {product[1]:.0f} "
        f"| Revenue: {product[2]:.2f}"
    )

print("\nPREVIOUS PERIOD COMPARISON")
print(f"Previous Period   : {previous_start} to {previous_end}")
print(f"Revenue Change    : {revenue_change:+.2f}%")
print(f"Orders Change     : {orders_change:+.2f}%")

print("=" * 50)
```

```python
# ==========================================
# PART 5: EDGE CASE HANDLING
# ==========================================

import sqlite3

# Connect to SQLite database
connection = sqlite3.connect("ecommerce.db")


# Test 1: order_items has an order_id not present in orders
def test_invalid_order_id():
    query = """
    SELECT COUNT(*)
    FROM order_items oi
    LEFT JOIN orders o
        ON oi.order_id = o.order_id
    WHERE o.order_id IS NULL;
    """

    result = connection.execute(query).fetchone()[0]

    if result == 0:
        print("Test 1 PASSED: No invalid order_id found.")
    else:
        print(f"Test 1: {result} invalid order_id(s) found.")


# Test 2: Check whether discount_percent > 100

def test_discount_greater_than_100():
    # Check which table contains discount_percent
    tables = connection.execute("""
        SELECT name
        FROM sqlite_master
        WHERE type = 'table';
    """).fetchall()

    found = False

    for table in tables:
        table_name = table[0]

        columns = connection.execute(
            f"PRAGMA table_info({table_name})"
        ).fetchall()

        column_names = [column[1] for column in columns]

        if "discount_percent" in column_names:
            found = True

            query = f"""
            SELECT COUNT(*)
            FROM {table_name}
            WHERE discount_percent > 100;
            """

            result = connection.execute(query).fetchone()[0]

            if result == 0:
                print("Test 2 PASSED: No discount greater than 100% found.")
            else:
                print(
                    f"Test 2: {result} record(s) have discount_percent > 100."
                )

            break

    if not found:
       print("Test 2: Not Applicable (discount_percent column is not available in this dataset).")


test_discount_greater_than_100()


# Test 3: quantity is 0
def test_zero_quantity():
    query = """
    SELECT COUNT(*)
    FROM order_items
    WHERE quantity = 0;
    """

    result = connection.execute(query).fetchone()[0]

    if result == 0:
        print("Test 3 PASSED: No order has quantity = 0.")
    else:
        print(f"Test 3: {result} record(s) have quantity = 0.")


# Test 4: order_date is in the future
def test_future_order_date():
    query = """
    SELECT COUNT(*)
    FROM orders
    WHERE DATE(order_date) > DATE('now');
    """

    result = connection.execute(query).fetchone()[0]

    if result == 0:
        print("Test 4 PASSED: No future order_date found.")
    else:
        print(f"Test 4: {result} order(s) have a future order_date.")


# Run all tests
print("EDGE CASE TEST RESULTS")
print("=" * 50)

test_invalid_order_id()
test_discount_greater_than_100()
test_zero_quantity()
test_future_order_date()
```
