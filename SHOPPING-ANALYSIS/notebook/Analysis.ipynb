## Handling Missing Values

Missing values are identified and replaced with appropriate values.

```python
# Count missing values in each column
combined_df.isnull().sum()
```

```python
# Fill missing values in numeric and text columns
combined_df["discount"] = combined_df["discount"].fillna(0)

combined_df["seller_name"] = combined_df["seller_name"].fillna("Unknown")

combined_df["seller_information"] = combined_df["seller_information"].fillna("Not Available")

combined_df["what_customers_said"] = combined_df["what_customers_said"].fillna("No Review")

combined_df["videos"] = combined_df["videos"].fillna("No Video")

combined_df["variations"] = combined_df["variations"].fillna("No Variation")
```

```python
# Verify that all missing values have been handled
combined_df.isnull().sum()
```

## Data Selection and Filtering

In this step, specific columns are selected from the dataset and rows are filtered based on certain conditions. This helps focus on relevant data for analysis.

```python
# Display only selected columns
combined_df[["title", "rating", "category"]]
```

```python
# Display products having rating greater than 4
combined_df[combined_df["rating"] > 4]
```

## Removing Duplicate Records

In this step, duplicate records in the dataset are identified and removed to ensure data quality and avoid redundant information during analysis.

```python
# Count duplicate rows in the dataset
combined_df.duplicated().sum()
```

```python
# Remove duplicate rows from the dataset
combined_df = combined_df.drop_duplicates()
```

## Creating Derived Columns

In this step, two new columns are created using existing columns in the dataset.

- `discount_amount` represents the discount value calculated from the initial price and discount percentage.
- `total_amount` represents the final product price after subtracting the discount amount from the initial price.

```python
# Create a new column 'discount_amount' that calculates the discount value
# using the formula: Initial Price × Discount Percentage / 100
combined_df["discount_amount"] = (
    combined_df["initial_price"] * combined_df["discount"] / 100
)
```

```python
# Display the first 5 rows of initial price, discount percentage,
# and calculated discount amount to verify the new column

combined_df[["initial_price", "discount", "discount_amount"]].head()
```

```python
# Create a derived column 'total_amount' by subtracting the discount
# amount from the initial product price
combined_df["total_amount"] = combined_df["initial_price"] - (
    combined_df["initial_price"] * combined_df["discount"] / 100
)
```

```python
# Display the first 5 rows of initial price, discount percentage,
# final price, and calculated total amount to verify the new column
combined_df[["initial_price", "discount", "final_price", "total_amount"]].head()
```

## Saving the Cleaned Dataset

After cleaning the data and creating derived columns, the processed dataset is saved as a new CSV file for future analysis and use.

```python
# Save the cleaned dataset as a new CSV file

combined_df.to_csv("cleaned_dataset.csv", index=False)
```
